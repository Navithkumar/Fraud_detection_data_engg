```python
from pathlib import Path
import pandas as pd
```

```python
DATA_DIR = Path("../data/raw")
FILES = [
    "train_transaction.csv",
    "train_identity.csv",
]


def inspect_csv(filename: str) -> None:
    path = DATA_DIR / filename
    print(path)

    if not path.exists():
        print(f"\n[ERROR] File not found: {path}")
        return  

    print(f"\n{'=' * 80}")
    print(f"FILE: {filename}")
    print(f"SIZE: {path.stat().st_size / (1024**2):.2f} MB")
    print("=" * 80)

    # Read a small sample first to inspect structure without
    # loading the entire dataset into memory.
    sample = pd.read_csv(path, nrows=1000)

    print("\nColumns and inferred data types:")
    print(sample.dtypes.to_string())

    print("\nSample null percentages:")
    print(
        (sample.isna().mean() * 100)
        .round(2)
        .sort_values(ascending=False)
        .head(20)
        .to_string()
    )

    print("\nFirst five rows:")
    print(sample.head().to_string())

    print(f"\nSample rows inspected: {len(sample):,}")
    print(f"Columns found: {len(sample.columns):,}")


def main() -> None:
    for filename in FILES:
        inspect_csv(filename)


if __name__ == "__main__":
    main()
```
